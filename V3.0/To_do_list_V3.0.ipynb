Notlar & İpuçları
Veri Tutarlılığı: Eğitim sırasında desteleri deterministik RNG seed’leriyle üret, yeniden üretilebilirlik sağlar.
Model Boyutu: Kart bilgisi yüksek boyutlu değil; MLP’ler yeterli. LSTM eklemeyi ancak uzun hafıza gerekiyorsa düşün.
Risk Ölçümleri: Bahis ajanı için Sharpe Ratio / Risk of Ruin metriklerini W&B’de logla.
Simülasyon Hızı: NumPy vektörleştirme + Numba/JIT ile blackjack_engine hızlandır, aksi takdirde 50 M el günler sürebilir.
Test Kapsamı: Her faz bittiğinde pytest --cov ≥ 85 % hedefle.
